```python
from pydantic import BaseModel, ConfigDict
```

```python
class User(BaseModel):
    id:int
    name:str="Jesilda"

    model_config= ConfigDict(str_max_length=18)
```

```python
user = User(id='123')
```

```python
assert user.name == 'JaneDoe'  
assert user.id == 123  
assert isinstance(user.id, int)
```
