```python
class Employee:
  num_of_employees=0
  raise_amount=1.04
  def __init__(self,fname,lname,pay):
    self.first=fname
    self.last=lname
    self.pay=pay
    self.em3ail=fname+lname+"@gmail.com"
    Employee.num_of_employees+=1

  def fullname(self):
    return '{} {}'.format(self.first,self.last)

  def apply_raise(self):
    self.pay=int(self.pay * self.raise_amount)
```

The Exercise: Notification Service for a Document Pipeline

Scenario: You're building a service that processes documents (think: an ADR report coming into PharmaWatch) and needs to notify someone when processing finishes — sometimes via email, sometimes via Slack, sometimes both, and this should be swappable/testable.

Requirements

Build these pieces using composition + dependency injection, not inheritance:

Notifier implementations — at least two classes that each expose a .send(message: str) -> bool method:

EmailNotifier (can just print f"EMAIL: {message}" — no real email needed)

SlackNotifier (prints f"SLACK: {message}")

They should not share a base class. Just make sure they both have .send(). (This is duck typing — revisit that term from the Real Python article if it's fuzzy.)

DocumentProcessor class:

__init__ should accept a notifier as a parameter (injected, not created internally)

Has a .process(document_text: str) -> dict method that does some fake "processing" (e.g. word count, or just .upper() the text — doesn't matter, keep it simple)


After processing, calls self.notifier.send(...) with a summary message

A FakeNotifier for testing:


Instead of printing, it should just store sent messages in a list (self.sent_messages = [])

Write a quick test (a plain function with assert, no pytest needed) that proves DocumentProcessor calls the notifier correctly — without touching 

EmailNotifier or SlackNotifier at all

The runtime-swap test:

Create one DocumentProcessor with an EmailNotifier

Process a document

Then swap .notifier to a SlackNotifier on the same object, mid-program

Process another document

Confirm the output changes accordingly
