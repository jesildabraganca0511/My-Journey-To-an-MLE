```python
class Employee:
  def __init__(self,fname,lname,pay):
    self.first=fname
    self.last=lname
    self.pay=pay
    self.email=fname+lname+"@gmail.com"
  def fullname(self):
    return '{} {}'.format(self.first,self.last)
```

```python
emp1=Employee("jess","braganca",2300)
emp2=Employee("jessy","braganca",2300)
```

```python
print(emp1.fullname())
print(emp2.fullname())
```
