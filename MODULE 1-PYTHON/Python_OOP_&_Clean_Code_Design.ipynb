```python
class Employee:
  num_of_employees=0
  raise_amount=1.04
  def __init__(self,fname,lname,pay):
    self.first=fname
    self.last=lname
    self.pay=pay
    self.em3ail=fname+lname+"@gmail.com"
    Employee.num_of_employees+=1

  def fullname(self):
    return '{} {}'.format(self.first,self.last)

  def apply_raise(self):
    self.pay=int(self.pay * self.raise_amount)
```

```python
emp1=Employee("jess","braganca",2300)
emp2=Employee("jessy","braganca",2300)
```

```python
print(emp1.fullname())
print(emp2.fullname())
```

```python
print(emp1.apply_raise())
```
